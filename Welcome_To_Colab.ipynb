```
import pandas as pd
import numpy as np
students = pd.DataFrame({
 "StudentID": [101,102,103,104,105,106,107,108,109,110,110,111],
 "Name": ["Ali","Sara","Ahmed","Hina","Usman","Fatima",
 "Bilal","Ayesha","Hamza","Zara","Zara","Danish"],
 "Gender": ["Male","Female","male","FEMALE","M","Female",
 "Male","female","Male","F","F","Male"],
 "Department": ["CS","SE","CS","AI","se","CS","AI","CS","SE","AI","AI","CS"],
 "Age": [20,21,22,20,np.nan,23,21,22,20,21,21,35]
})
attendance = pd.DataFrame({
 "StudentID": [101,102,103,104,105,106,107,108,109,110,111],
 "ClassesHeld": [40]*11,
 "ClassesAttended": [37,32,22,39,31,35,18,36,28,41,25]
})
results = pd.DataFrame({
 "StudentID": [101,102,103,104,105,106,107,108,109,110,111],
 "Quiz1": [8,7,5,9,np.nan,8,4,9,6,7,5],
 "Quiz2": [9,8,6,9,7,8,5,10,7,8,np.nan],
 "Assignment": [18,16,13,20,15,17,10,19,14,18,12],
 "Midterm": [25,22,16,28,20,24,14,27,19,23,15],
 "Final": [35,31,22,38,28,34,20,37,27,33,18]
})
students.to_csv("students.csv", index=False)
attendance.to_csv("attendance.csv", index=False)
results.to_csv("results.csv", index=False)
print("Files created successfully")
```

```
df_students = pd.read_csv('students.csv')
df_attendance = pd.read_csv('attendance.csv')
df_results = pd.read_csv('results.csv')

print("--- Missing Values in Students ---")
print(df_students.isnull().sum())

print("\n--- Missing Values in Attendance ---")
print(df_attendance.isnull().sum())

print("\n--- Missing Values in Results ---")
print(df_results.isnull().sum())
```

```
# Display complete duplicated records based on StudentID
duplicate_records = df_students[df_students.duplicated(subset="StudentID", keep=False)]
print("Duplicated Student records based on StudentID:")
print(duplicate_records)
```

```
# A student cannot attend more classes than were held
invalid_attendance = df_attendance[df_attendance["ClassesAttended"] > df_attendance["ClassesHeld"]]
print("Invalid Attendance Records:")
print(invalid_attendance)
```

```
# Create clean copy
clean_students = df_students.copy()

# Remove duplicate StudentID entries
clean_students = clean_students.drop_duplicates(subset="StudentID")

# Verify uniqueness
print(f"Number of unique StudentIDs: {clean_students['StudentID'].nunique()}")
print(f"Remaining duplicate row count: {clean_students.duplicated(subset='StudentID').sum()}")
```

```
# Define mapping dictionary
gender_mapping = {
    'Male': 'Male', 'male': 'Male', 'M': 'Male',
    'Female': 'Female', 'female': 'Female', 'F': 'Female', 'FEMALE': 'Female'
}

# Standardize values
clean_students['Gender'] = clean_students['Gender'].replace(gender_mapping)

# Display unique values to verify
print("Unique values in Gender column:", clean_students['Gender'].unique())
```

```
# Force upper-case normalization
clean_students['Department'] = clean_students['Department'].str.upper()

# Display unique values to verify
print("Unique values in Department column:", clean_students['Department'].unique())
```

```
# Calculate the median age dynamically from data
median_age = clean_students['Age'].median()

# Fill missing items
clean_students['Age'] = clean_students['Age'].fillna(median_age)

# Verify no missing values remain
print(f"Calculated Median Age: {median_age}")
print(f"Missing values remaining in Age: {clean_students['Age'].isnull().sum()}")
```

```
# Create clean copy
clean_results = df_results.copy()

# Fill missing values using individual column medians
clean_results['Quiz1'] = clean_results['Quiz1'].fillna(clean_results['Quiz1'].median())
clean_results['Quiz2'] = clean_results['Quiz2'].fillna(clean_results['Quiz2'].median())

# Confirm cleaning
print("Missing values remaining in Quiz1:", clean_results['Quiz1'].isnull().sum())
print("Missing values remaining in Quiz2:", clean_results['Quiz2'].isnull().sum())
```

```
# Create clean copy
clean_attendance = df_attendance.copy()

# Correction Rule: If ClassesAttended exceeds ClassesHeld, cap it at ClassesHeld assuming a manual typo.
clean_attendance.loc[clean_attendance['ClassesAttended'] > clean_attendance['ClassesHeld'], 'ClassesAttended'] = clean_attendance['ClassesHeld']

# Verify correction
print(clean_attendance)
```

```
# Calculate percentage and round safely to 2 decimals
clean_attendance['AttendancePercentage'] = ((clean_attendance['ClassesAttended'] / clean_attendance['ClassesHeld']) * 100).round(2)
print(clean_attendance[['StudentID', 'AttendancePercentage']].head())
```

```
# Sum components together
clean_results['QuizTotal'] = clean_results['Quiz1'] + clean_results['Quiz2']
clean_results['TotalMarks'] = clean_results['QuizTotal'] + clean_results['Assignment'] + clean_results['Midterm'] + clean_results['Final']
clean_results['Percentage'] = (clean_results['TotalMarks'] / 100) * 100

print(clean_results[['StudentID', 'QuizTotal', 'TotalMarks', 'Percentage']].head())
```

```
# Define conditional function logic
def assign_performance_tier(pct):
    if pct >= 85:
        return "Excellent"
    elif pct >= 70:
        return "Good"
    elif pct >= 60:
        return "Average"
    else:
        return "Needs Improvement"

# Map to dataframe column
clean_results['Performance'] = clean_results['Percentage'].apply(assign_performance_tier)
print(clean_results[['StudentID', 'Percentage', 'Performance']].head())
```

```
# Save clean DataFrames out to disk without indices
clean_students.to_csv("students_clean.csv", index=False)
clean_attendance.to_csv("attendance_clean.csv", index=False)
clean_results.to_csv("results_clean.csv", index=False)

print("All cleaned files exported successfully!")
```

```
# Compile dataset analytics data row by row
quality_data = [
    {
        "Dataset": "Students",
        "Rows": clean_students.shape[0],
        "Columns": clean_students.shape[1],
        "MissingValues": clean_students.isnull().sum().sum(),
        "DuplicateIDs": clean_students.duplicated(subset="StudentID").sum()
    },
    {
        "Dataset": "Attendance",
        "Rows": clean_attendance.shape[0],
        "Columns": clean_attendance.shape[1],
        "MissingValues": clean_attendance.isnull().sum().sum(),
        "DuplicateIDs": clean_attendance.duplicated(subset="StudentID").sum()
    },
    {
        "Dataset": "Results",
        "Rows": clean_results.shape[0],
        "Columns": clean_results.shape[1],
        "MissingValues": clean_results.isnull().sum().sum(),
        "DuplicateIDs": clean_results.duplicated(subset="StudentID").sum()
    }
]

# Generate DataFrame and export as CSV
df_quality_report = pd.DataFrame(quality_data)
df_quality_report.to_csv("data_quality_report.csv", index=False)

print("\n--- Data Quality Summary Report ---")
print(df_quality_report)
```
